## Алхам 2: Хүлээгдэж буй ашиг

```python
def expected_utility(outcomes):
    total = 0
    for p, u in outcomes:
        total += p
    if abs(total - 1) > 1e-9:
        raise ValueError("Магадлалын нийлбэр 1 биш байна")

    eu = 0
    for p, u in outcomes:
        eu += p * u
    return eu
```

```python
study = [(0.9, 80), (0.1, 10)]
play = [(0.4, 100), (0.6, 0)]
a = [(1.0, 100000)]
b = [(0.5, 250000), (0.5, 0)]

print("Шалгалтад бэлдэх:", expected_utility(study))
print("Тоглох:", expected_utility(play))
print("А: шууд мөнгө:", expected_utility(a))
print("Б: зоос шидэх:", expected_utility(b))
```

```python
try:
    expected_utility([(0.5, 10), (0.4, 20)])
except ValueError as e:
    print("Алдаа:", e)
```
