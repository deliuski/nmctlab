## 0) Dataset унших ба ерөнхий шалгалт

Өмнөх lab-уудын адил **StudentsPerformance.csv** dataset-ийг ашиглана.


```python
import pandas as pd

df = pd.read_csv("StudentsPerformance.csv")
df.head()
```

## 1) Feature Construction – Шинэ feature үүсгэх

Raw score-ууд заримдаа хангалттай мэдээлэл өгдөггүй.  
Тиймээс шинэ feature үүсгэнэ.


### TASK 1: Average score

1. Math, Reading, Writing score-уудын **дундаж**-ийг ол.
2. `average_score` нэртэй шинэ column үүсгэ.


```python
# TODO: average_score үүсгэ
df['average_score'] = (df['math score'] + df['reading score'] + df['writing score']) / 3

df[['average_score']].head()
```

## 2) Feature Transformation – Discretization

Numeric feature-ийг categorical болгоё.


### TASK 2: Score discretization

Доорх ангиллыг ашиглана:

- average_score < 60 → Low  
- 60 ≤ average_score < 80 → Medium  
- average_score ≥ 80 → High  

`performance_level` нэртэй шинэ column үүсгэ.


```python
# TODO: discretization хийх
def score_to_level(x):
    if x < 60:
        return "Low"
    elif x < 80:
        return "Medium"
    else:
        return "High"

df['performance_level'] = df['average_score'].apply(score_to_level)
df[['average_score','performance_level']].head()
```

## 3) Масштаб (Scale) ба Normalization

Одоо масштабын асуудлыг **тооцуулж** харъя.


### TASK 3: Min–Max Normalization

1. Math score-ийн min, max утгыг ол.
2. Min–Max normalization томъёогоор `math_norm` column үүсгэ.

Томъёо:
(x − min) / (max − min)


```python
# TODO: Min–Max normalization
math_min = df['math score'].min()
math_max = df['math score'].max()

df['math_norm'] = (df['math score'] - math_min) / (math_max - math_min)

df[['math score','math_norm']].head()
```
